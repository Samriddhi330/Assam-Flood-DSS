```python
import os
import pandas as pd

# 1. Download CivicDataLab repository if not present
repo_path = 'flood-data-ecosystem-Assam'
if not os.path.exists(repo_path):
    print("Downloading CivicDataLab repository into Colab...\n")
    !git clone https://github.com/CivicDataLab/flood-data-ecosystem-Assam.git

# 2. Path to master dataset
dataset_path = 'flood-data-ecosystem-Assam/Sources/MASTER_VARIABLES.csv'

# 3. Read dataset
df = pd.read_csv(dataset_path)

print(f"✅ STEP 1 COMPLETE: Raw dataset loaded successfully!")
print(f"Total Rows: {df.shape[0]} | Total Columns: {df.shape[1]}\n")

print("--- FIRST 5 ROWS OF RAW DATA ---")
display(df.head())

print("\n--- RAW DATASET INFO & STRUCTURE ---")
df.info()
```

```python
# 1. Define essential columns for ML modeling
selected_columns = [
    'object_id', 'district', 'revenue_ci', 'timeperiod',
    # Hazard (XH)
    'sum_rain', 'max_rain', 'riverlevel_mean', 'distance_from_river', 'inundation_pct',
    # Exposure (XE)
    'sum_population', 'total_hhd', 'Crop_Area',
    # Vulnerability (XV)
    'road_length', 'health_centres_count', 'elevation_mean', 'slope_mean',
    # Raw Target
    'Population_affected_Total'
]

# 2. Filter dataset
df_clean = df[selected_columns].copy()

# 3. Calculate Target Variable: Affected Population Ratio (Y)
# Formula: Population_affected_Total / (sum_population + 1), bounded between 0 and 1
df_clean['affected_pop_ratio'] = (df_clean['Population_affected_Total'] / (df_clean['sum_population'] + 1)).clip(0, 1)

print(f"✅ STEP 2 COMPLETE: Selected 16 essential columns & calculated target variable!")
print(f"New Shape: {df_clean.shape[0]} Rows × {df_clean.shape[1]} Columns\n")

print("--- FIRST 5 ROWS OF SELECTED DATASET ---")
display(df_clean.head())

print("\n--- SELECTED DATASET INFO & STRUCTURE ---")
df_clean.info()
```

```python
import numpy as np

print("--- 1. MISSING (NULL) VALUE COUNT PER COLUMN ---")
nulls = df_clean.isnull().sum()
print(nulls)

print("\n--- 2. STATISTICAL SUMMARY OF NUMERICAL FEATURES ---")
display(df_clean.describe())

# Handle any missing or infinite values
num_cols = df_clean.select_dtypes(include=[np.number]).columns
df_clean[num_cols] = df_clean[num_cols].replace([np.inf, -np.inf], np.nan)
df_clean[num_cols] = df_clean[num_cols].fillna(df_clean[num_cols].median())

# Ensure physical values are non-negative
df_clean['sum_rain'] = df_clean['sum_rain'].clip(lower=0)
df_clean['max_rain'] = df_clean['max_rain'].clip(lower=0)
df_clean['sum_population'] = df_clean['sum_population'].clip(lower=0)
df_clean['distance_from_river'] = df_clean['distance_from_river'].clip(lower=0)

print("\n✅ STEP 3 COMPLETE: Missing values & physical bounds verified!")
```

```python
# Save final dataset locally in Colab
df_clean.to_csv('master_dataset.csv', index=False)

print(f"✅ STEP 4 COMPLETE: Final dataset saved as 'master_dataset.csv'!")
print(f"Final File Shape: {df_clean.shape[0]} Rows × {df_clean.shape[1]} Columns\n")

print("--- FINAL CLEAN DATASET PREVIEW (FIRST 5 ROWS) ---")
display(df_clean.head())

print("\n--- FINAL COLUMN NAMES LIST ---")
for idx, col in enumerate(df_clean.columns):
    print(f"Column {idx+1}: {col}")
```
