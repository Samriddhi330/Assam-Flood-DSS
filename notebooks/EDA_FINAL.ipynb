```python
import json

rdls_path = os.path.join(
    source_folder,
    "Sources",
    "DRIMS",
    "metadata",
    "rdls.json"
)

with open(rdls_path, "r", errors="ignore") as f:
    rdls = json.load(f)

print(type(rdls))

# Search recursively for our variables
def find_variable(obj, target_names, path="root"):
    if isinstance(obj, dict):
        for key, value in obj.items():
            current_path = f"{path}.{key}"

            if str(key).lower() in [x.lower() for x in target_names]:
                print("\nFOUND KEY:", current_path)
                print(value)

            find_variable(value, target_names, current_path)

    elif isinstance(obj, list):
        for i, item in enumerate(obj):
            find_variable(item, target_names, f"{path}[{i}]")

find_variable(
    rdls,
    ["Population_affected_Total", "sum_population"]
)
```

```python
# Check temporal persistence of population impact

temp_df = df[
    ["object_id", "timeperiod", "Population_affected_Total"]
].copy()

# Convert timeperiod to proper datetime
temp_df["date"] = pd.to_datetime(
    temp_df["timeperiod"],
    format="%Y_%m"
)

# Sort spatially and temporally
temp_df = temp_df.sort_values(
    ["object_id", "date"]
)

# Binary impact indicator
temp_df["impact"] = (
    temp_df["Population_affected_Total"] > 0
).astype(int)

# Previous month's impact for the same Revenue Circle
temp_df["previous_impact"] = (
    temp_df.groupby("object_id")["impact"]
    .shift(1)
)

# Cases where current month has impact
current_impact = temp_df["impact"] == 1

# Consecutive impact:
# current month affected AND previous month affected
consecutive = (
    current_impact &
    (temp_df["previous_impact"] == 1)
)

print("Total non-zero impact observations:",
      current_impact.sum())

print("Impact observations preceded by another impacted month:",
      consecutive.sum())

print(
    "Percentage of impacted observations preceded by impact:",
    round(
        100 * consecutive.sum() / current_impact.sum(),
        2
    ),
    "%"
)
```

```python
# Analyze month-to-month transitions in population impact

transition_df = temp_df.copy()

# Remove the first month for each Revenue Circle
transition_df = transition_df.dropna(
    subset=["previous_impact"]
)

# Create transition labels
transition_df["transition"] = (
    transition_df["previous_impact"].astype(int).astype(str)
    + " -> "
    + transition_df["impact"].astype(int).astype(str)
)

print("Month-to-month impact transitions:\n")
print(
    transition_df["transition"]
    .value_counts()
    .sort_index()
)

print("\n" + "="*70)

# Conditional probability of current impact
# given previous month's state

prob_previous_impact = transition_df.loc[
    transition_df["previous_impact"] == 1,
    "impact"
].mean()

prob_previous_no_impact = transition_df.loc[
    transition_df["previous_impact"] == 0,
    "impact"
].mean()

print(
    "P(Current impact | Previous month impacted):",
    round(prob_previous_impact, 4)
)

print(
    "P(Current impact | Previous month not impacted):",
    round(prob_previous_no_impact, 4)
)

print("\nDifference:")
print(
    round(
        prob_previous_impact - prob_previous_no_impact,
        4
    )
)
```

```python


rain_df = df[
    [
        "object_id",
        "timeperiod",
        "Population_affected_Total",
        "sum_rain",
        "max_rain",
        "mean_rain"
    ]
].copy()

# Convert timeperiod to datetime
rain_df["date"] = pd.to_datetime(
    rain_df["timeperiod"],
    format="%Y_%m"
)

# Sort by Revenue Circle and time
rain_df = rain_df.sort_values(
    ["object_id", "date"]
)

# Binary impact
rain_df["impact"] = (
    rain_df["Population_affected_Total"] > 0
).astype(int)

# Previous-month rainfall
for col in ["sum_rain", "max_rain", "mean_rain"]:
    rain_df[f"previous_{col}"] = (
        rain_df.groupby("object_id")[col].shift(1)
    )

# Remove first month of each spatial unit
rain_df = rain_df.dropna(
    subset=[
        "previous_sum_rain",
        "previous_max_rain",
        "previous_mean_rain"
    ]
)

print("Current-month rainfall statistics by impact:\n")

print(
    rain_df.groupby("impact")[
        ["sum_rain", "max_rain", "mean_rain"]
    ].mean()
)

print("\n" + "="*70)

print("Previous-month rainfall statistics by CURRENT impact:\n")

print(
    rain_df.groupby("impact")[
        [
            "previous_sum_rain",
            "previous_max_rain",
            "previous_mean_rain"
        ]
    ].mean()
)
```

```python


spatial_impact = (
    df.groupby("object_id")
      .agg(
          total_months=("Population_affected_Total", "size"),
          impacted_months=(
              "Population_affected_Total",
              lambda x: (x > 0).sum()
          ),
          max_affected_population=(
              "Population_affected_Total",
              "max"
          )
      )
)

spatial_impact["impact_rate"] = (
    spatial_impact["impacted_months"]
    / spatial_impact["total_months"]
)

print("Impact-rate statistics across Revenue Circles:\n")
print(
    spatial_impact["impact_rate"].describe()
)

print("\n" + "="*70)

print("Revenue Circles with highest impact rates:\n")

print(
    spatial_impact
    .sort_values("impact_rate", ascending=False)
    .head(15)
)

print("\n" + "="*70)

print("Revenue Circles with zero reported impact:\n")

print(
    (spatial_impact["impacted_months"] == 0).sum(),
    "out of",
    len(spatial_impact)
)
```

```python


# Sort observations chronologically within each Revenue Circle
df_sorted = df.sort_values(
    ["object_id", "timeperiod"]
).copy()

# Current-month impact indicator
df_sorted["current_impact"] = (
    df_sorted["Population_affected_Total"] > 0
).astype(int)

# Next month's impact for the same Revenue Circle
df_sorted["next_month_impact"] = (
    df_sorted
    .groupby("object_id")["current_impact"]
    .shift(-1)
)

# Remove the final month of each Revenue Circle
prediction_data = df_sorted.dropna(
    subset=["next_month_impact"]
).copy()

prediction_data["next_month_impact"] = (
    prediction_data["next_month_impact"].astype(int)
)

print("Observations available for one-month-ahead analysis:",
      len(prediction_data))

print("\nNext-month impact distribution:")
print(
    prediction_data["next_month_impact"]
    .value_counts()
)

print("\nNext-month impact rate:")
print(
    prediction_data["next_month_impact"].mean()
)

print("\n" + "="*70)

# Relationship between current and next-month impact
transition = pd.crosstab(
    prediction_data["current_impact"],
    prediction_data["next_month_impact"]
)

print("Current impact → Next-month impact:")
print(transition)

print("\n" + "="*70)

print(
    "P(Next-month impact | Current impact):",
    prediction_data.loc[
        prediction_data["current_impact"] == 1,
        "next_month_impact"
    ].mean()
)

print(
    "P(Next-month impact | Current no impact):",
    prediction_data.loc[
        prediction_data["current_impact"] == 0,
        "next_month_impact"
    ].mean()
)
```

```python


rainfall_prediction = prediction_data.groupby(
    "next_month_impact"
)[
    [
        "sum_rain",
        "max_rain",
        "mean_rain"
    ]
].mean()

print("Current-month rainfall by NEXT-month impact:\n")
print(rainfall_prediction)

print("\n" + "="*70)

# Compare rainfall between the two groups
for col in ["sum_rain", "max_rain", "mean_rain"]:

    impact_mean = prediction_data.loc[
        prediction_data["next_month_impact"] == 1,
        col
    ].mean()

    no_impact_mean = prediction_data.loc[
        prediction_data["next_month_impact"] == 0,
        col
    ].mean()

    print(f"\n{col}")

    print("  Next month impacted     :", impact_mean)
    print("  Next month not impacted :", no_impact_mean)

    if no_impact_mean != 0:
        print(
            "  Ratio:",
            impact_mean / no_impact_mean
        )
```

```python

availability = []

for col in prediction_features:
    if col in df.columns:
        nonzero = (df[col].fillna(0) != 0).sum()
        missing = df[col].isna().sum()
        unique = df[col].nunique(dropna=True)

        availability.append({
            'variable': col,
            'nonzero': nonzero,
            'missing': missing,
            'unique': unique,
            'coverage_pct': round(
                100 * (len(df) - missing) / len(df), 2
            )
        })
    else:
        availability.append({
            'variable': col,
            'nonzero': 'NOT FOUND',
            'missing': 'NOT FOUND',
            'unique': 'NOT FOUND',
            'coverage_pct': 'NOT FOUND'
        })

availability_df = pd.DataFrame(availability)

display(availability_df)
```

```python


df_pred = df.copy()

# Convert YYYY_MM into an actual date
df_pred['date'] = pd.to_datetime(
    df_pred['timeperiod'].astype(str),
    format='%Y_%m'
)

# Sort by Revenue Circle and time
df_pred = df_pred.sort_values(
    ['object_id', 'date']
).reset_index(drop=True)

# ------------------------------------------------------------
# NEXT MONTH'S IMPACT
# ------------------------------------------------------------

df_pred['next_month_impact'] = (
    df_pred
    .groupby('object_id')['impact']
    .shift(-1)
)

# Next month's date
df_pred['next_month_date'] = (
    df_pred
    .groupby('object_id')['date']
    .shift(-1)
)

# ------------------------------------------------------------
# VERIFY THAT IT IS ACTUALLY THE IMMEDIATE NEXT MONTH
# ------------------------------------------------------------

month_difference = (
    (df_pred['next_month_date'].dt.year -
     df_pred['date'].dt.year) * 12
    +
    (df_pred['next_month_date'].dt.month -
     df_pred['date'].dt.month)
)

df_pred['valid_next_month'] = (
    month_difference == 1
)

# Keep only genuine one-month-ahead observations
df_pred = df_pred[
    df_pred['valid_next_month']
].copy()

# ------------------------------------------------------------
# SUMMARY
# ------------------------------------------------------------

print("=" * 70)
print("ONE-MONTH-AHEAD PREDICTION DATASET")
print("=" * 70)

print("Total observations:", len(df_pred))

print(
    "Revenue Circles:",
    df_pred['object_id'].nunique()
)

print(
    "Time range:",
    df_pred['date'].min(),
    "→",
    df_pred['date'].max()
)

print("\nNext-month impact distribution:")
print(
    df_pred['next_month_impact']
    .value_counts()
    .sort_index()
)

print("\nNext-month impact proportion:")
print(
    df_pred['next_month_impact'].mean()
)

print("\nValid next-month observations:")
print(
    df_pred['valid_next_month'].value_counts()
)
```
